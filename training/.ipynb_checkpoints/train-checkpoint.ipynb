```python
def visualize_desktop_samples():
    dataset_path = PROJECT_PATH / "dataset" / "train"
    images_path = dataset_path / "images"
    
    image_files = list(images_path.glob("*.*"))[:6]
    
    if not image_files:
        print("❌ No images found in training set!")
        return
    
    fig, axes = plt.subplots(2, 3, figsize=(15, 10))
    fig.suptitle('Sample Training Images from Desktop', fontsize=16)
    
    for i, img_path in enumerate(image_files):
        if i >= 6:
            break
            
        img = cv2.imread(str(img_path))
        img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
        
        ax = axes[i//3, i%3]
        ax.imshow(img)
        ax.set_title(f"Image: {img_path.name}")
        ax.axis('off')
    
    plt.tight_layout()
    plt.show()

visualize_desktop_samples()
```

```python
# Initialize YOLOv11 model
print("🚀 Initializing YOLOv11 model for Desktop...")

model = YOLO('yolo11s.pt')  # Using small version for good balance of speed and accuracy
print("✅ Model loaded successfully!")

# Verify GPU availability
print(f"🎯 GPU Available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"🎯 GPU Device: {torch.cuda.get_device_name(0)}")
    print(f"🎯 GPU Memory: {torch.cuda.get_device_properties(0).total_memory / 1024**3:.1f} GB")

# Training configuration for Desktop - GPU optimized
training_config = {
    'data': str(PROJECT_PATH / "dataset" / "dataset.yaml"),
    'epochs': 50,
    'imgsz': 640,
    'batch': 20,
    'name': 'desktop_crop_weed',
    'patience': 20,
    'device': 'cuda',  # Use 'cuda' instead of 0 to explicitly use GPU
    'optimizer': 'AdamW',
    'lr0': 0.001,
    'lrf': 0.01,
    'momentum': 0.9,
    'weight_decay': 0.0005,
    'warmup_epochs': 3,
    'amp': True,  # Automatic Mixed Precision
}

print("📊 Training Configuration:")
print("⚡ Optimized for RTX 3050 6GB GPU")
for key, value in training_config.items():
    print(f"  {key}: {value}")

# Start training
print("🏋️ Starting training on Desktop...")
print("⏰ Estimated time: 25-35 minutes with your RTX 3050")
print("📊 Monitoring training progress in the output below...")

try:
    results = model.train(**training_config)
    print("✅ Training completed successfully!")
    
except Exception as e:
    print(f"❌ Training error: {e}")
    print("Trying with explicit device index...")
    
    # Fallback configuration with explicit device
    fallback_config = {
        'data': str(PROJECT_PATH / "dataset" / "dataset.yaml"),
        'epochs': 50,
        'imgsz': 640,
        'batch': 16,
        'device': 0,  # Explicit device index
        'amp': True,
    }
    
    try:
        results = model.train(**fallback_config)
        print("✅ Training completed with batch size 16!")
    except Exception as e2:
        print(f"❌ Still failing: {e2}")
        print("Trying with CPU fallback...")
        
        # Final fallback - CPU
        final_config = {
            'data': str(PROJECT_PATH / "dataset" / "dataset.yaml"),
            'epochs': 50,
            'imgsz': 640,
            'batch': 8,
            'device': 'cpu',  # CPU fallback
        }
        
        results = model.train(**final_config)
        print("✅ Training completed on CPU!")
```

```python

def predict_yield_desktop(image_path, model):
    """Predict crop and weed detection on Desktop"""
    try:
        # Run inference
        results = model(image_path, conf=0.25)
        result = results[0]  # take first result
        
        if result.boxes is None or len(result.boxes) == 0:
            return {
                'success': True,
                'crop_count': 0,
                'weed_count': 0,
                'weed_proximity_score': 0,
                'predicted_yield': 0,
                'yield_percentage': 0,
                'image': result.plot()
            }
        
        crop_count = 0
        weed_count = 0
        weed_proximity_score = 0
        
        # Count detections
        for box in result.boxes:
            class_id = int(box.cls)
            
            if class_id == 0:  # crop
                crop_count += 1
            elif class_id == 1:  # weed
                weed_count += 1
                
                # Calculate proximity
                bbox = box.xyxy[0].cpu().numpy()
                weed_center = np.array([(bbox[0] + bbox[2]) / 2,
                                        (bbox[1] + bbox[3]) / 2])
                
                min_distance = float('inf')
                for other_box in result.boxes:
                    if int(other_box.cls) == 0:  # crop
                        crop_bbox = other_box.xyxy[0].cpu().numpy()
                        crop_center = np.array([(crop_bbox[0] + crop_bbox[2]) / 2,
                                                (crop_bbox[1] + crop_bbox[3]) / 2])
                        distance = np.linalg.norm(weed_center - crop_center)
                        min_distance = min(min_distance, distance)
                
                if min_distance < float('inf'):
                    weed_proximity_score += 1 / (min_distance + 1)
        
        # Yield calculation
        base_yield = crop_count * 10 if crop_count > 0 else 0
        yield_reduction = weed_proximity_score * 2
        predicted_yield = max(0, base_yield - yield_reduction)
        yield_percentage = (predicted_yield / base_yield * 100) if base_yield > 0 else 0
        
        return {
            'image': result.plot(),  # numpy image (BGR)
            'crop_count': crop_count,
            'weed_count': weed_count,
            'weed_proximity_score': weed_proximity_score,
            'predicted_yield': predicted_yield,
            'yield_percentage': yield_percentage,
            'success': True
        }
        
    except Exception as e:
        print(f"❌ Prediction error: {e}")
        return {'success': False, 'error': str(e)}


# 🧪 Test prediction
print("🧪 Testing prediction on Desktop...")
test_images_dir = PROJECT_PATH / "dataset" / "test" / "images"
test_images = list(test_images_dir.glob("*.*"))

if test_images:
    test_image = test_images[0]
    print(f"Testing with: {test_image.name}")
    
    prediction = predict_yield_desktop(str(test_image), model)
    
    if prediction and prediction['success']:
        print(f"✅ Prediction successful!")
        print(f"🌱 Crops: {prediction['crop_count']}")
        print(f"🌿 Weeds: {prediction['weed_count']}")
        print(f"📏 Weed proximity score: {prediction['weed_proximity_score']:.2f}")
        print(f"📈 Predicted yield: {prediction['predicted_yield']:.1f}")
        print(f"📊 Yield percentage: {prediction['yield_percentage']:.1f}%")
        
        # Show result (convert BGR → RGB for matplotlib)
        plt.figure(figsize=(10, 8))
        plt.imshow(cv2.cvtColor(prediction['image'], cv2.COLOR_BGR2RGB))
        plt.title(f"Desktop Prediction: {test_image.name}")
        plt.axis('off')
        plt.show()
    else:
        print("❌ Prediction failed!")
else:
    print("❌ No test images found!")
```
